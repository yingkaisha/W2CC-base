```python
import os
import numpy as np
import pandas as pd
import xarray as xr
from tqdm import tqdm
 
from scipy.stats import norm
from scipy.optimize import minimize
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# verified variables
vars_verif = [
    'Fairbanks_T2_1d_max',
    'Fairbanks_T2_1h_max',
    'Fairbanks_T2_30d_max',
    'Fairbanks_T2_7d_max',
    'Fort_Bragg_PRECT_1d_max',
    'Fort_Bragg_PRECT_3d_max',
    'Fort_Bragg_PRECT_5d_max',
    'Fort_Bragg_SPEI_09_mean',
    'Fort_Bragg_SPEI_09_min',
    'Fort_Bragg_SPEI_48_mean',
    'Fort_Bragg_SPEI_48_min',
    'Guam_SPEI_03_mean',
    'Guam_SPEI_03_min',
    'Guam_SPEI_48_mean',
    'Guam_SPEI_48_min',
    'Pituffik_FT_days',
    'Pituffik_MDD',
    'Yuma_PG_PRECT_1d_max',
    'Yuma_PG_PRECT_7d_max',
    'Yuma_PG_SPEI_24_mean',
    'Yuma_PG_SPEI_24_min',
    'Yuma_PG_SPEI_48_mean',
    'Yuma_PG_SPEI_48_min',
]

years_verif = np.arange(2000, 2026)
years_train = np.arange(1959, 2000)
data_dir = '/glade/derecho/scratch/ksha/EPRI_data/METRICS/'

ds_target = xr.open_zarr(os.path.join(data_dir, 'STN_ERA5_ALL.zarr'))[vars_verif]

# ds_target_hist for determine percentiles
ds_target_hist = ds_target.sel(valid_year=years_train)

# ds_target with dims of `(valid_year: 26)`, valid_year = init_year + lead_year, up to 2025
ds_target = ds_target.sel(valid_year=years_verif)[vars_verif]

ds_dress = xr.open_zarr('/glade/derecho/scratch/ksha/EPRI_data/PP_calib/BMA/Dress.zarr')
ds_dress = ds_dress.rename({'dressed_member': 'member'})
ds_dress_qm = xr.open_zarr('/glade/derecho/scratch/ksha/EPRI_data/PP_calib/BMA/Dress_QM.zarr')
ds_dress_qm = ds_dress_qm.rename({'dressed_member': 'member'})

ds_calib = xr.open_zarr('/glade/derecho/scratch/ksha/EPRI_data/PP_calib/BMA/Calib.zarr')

ds_dress_qm_mean = ds_dress_qm.mean('member')
ds_dress_qm_mean = ds_dress_qm_mean.rename({v: f"{v}_mu" for v in ds_dress_qm_mean.data_vars})

ds_dress_qm_std = ds_dress_qm.std('member')
ds_dress_qm_std = ds_dress_qm_std.rename({v: f"{v}_sigma" for v in ds_dress_qm_std.data_vars})

ds_calib_qm = xr.merge([ds_dress_qm_mean, ds_dress_qm_std])

# ds_dress_all with dims of `(init_year: 21, lead_year: 10, member: 50)` and 23 post-processed variables
ds_dress_all = xr.merge([ds_dress, ds_dress_qm])[vars_verif]

# ds_calib_all with dims of `(init_year: 21, lead_year: 10)` and 46 variables, which are the `_mu` and `_sigma` of each post-processed variables.
ds_calib_all = xr.merge([ds_calib, ds_calib_qm])
```

```python
# ============================================================================
# Config (assume vars_verif, ds_target, ds_target_hist, ds_dress_all,
# ds_calib_all, years_verif are already defined in the surrounding script)
# ============================================================================
N_REL_BINS    = 8   # reliability diagram forecast-probability bins
N_RANK_BINS   = None # set to n_member + 1 below
N_SPREAD_BINS = 5   # spread-skill bins
MIN_PER_BIN   = 5    # minimum samples for a bin to be reported
```

```python
# ============================================================================
# Helper: pair forecasts with targets by valid_year = init_year + lead_year
# ============================================================================
def gather_pairs(da_fc, da_obs, has_member=True):
    """Flatten (init_year, lead_year[, member]) forecast and matching obs
    into 1D arrays paired on valid_year. Returns (fc, obs) where fc is
    either (N,) or (N, n_member) and obs is (N,)."""
    init_years = da_fc['init_year'].values
    lead_years = da_fc['lead_year'].values
    obs_years  = da_obs['valid_year'].values
 
    II, LL = np.meshgrid(init_years, lead_years, indexing='ij')
    valid  = (II + LL).ravel()                              # (n_init * n_lead,)
    fc_flat = (da_fc.values.reshape(-1, da_fc.sizes['member'])
               if has_member else da_fc.values.ravel())
 
    in_mask = np.isin(valid, obs_years)
    if not in_mask.any():
        return (np.empty((0, da_fc.sizes['member'])) if has_member
                else np.empty(0)), np.empty(0)
 
    obs_lookup = da_obs.sel(valid_year=valid[in_mask]).values
 
    if has_member:
        fc_out = fc_flat[in_mask, :]
    else:
        fc_out = fc_flat[in_mask]
 
    finite_fc  = (np.all(np.isfinite(fc_out), axis=1) if has_member
                  else np.isfinite(fc_out))
    finite_obs = np.isfinite(obs_lookup)
    keep = finite_fc & finite_obs
    return (fc_out[keep] if not has_member else fc_out[keep, :]), obs_lookup[keep]
```

```python
# ============================================================================
# 1. Reliability diagram
# ============================================================================
def reliability_curve(prob_fc, obs_binary, n_bins=N_REL_BINS):
    """Bin forecast probabilities and compute observed frequency per bin.
    Returns: bin_centers (n_bins,), fc_freq, obs_freq, count (all n_bins,)."""
    edges = np.linspace(0.0, 1.0, n_bins + 1)
    centers = 0.5 * (edges[:-1] + edges[1:])
 
    fc_freq  = np.full(n_bins, np.nan)
    obs_freq = np.full(n_bins, np.nan)
    count    = np.zeros(n_bins, dtype=np.int64)
 
    if prob_fc.size == 0:
        return centers, fc_freq, obs_freq, count
 
    idx = np.clip(np.digitize(prob_fc, edges) - 1, 0, n_bins - 1)
    for k in range(n_bins):
        m = (idx == k)
        count[k] = int(m.sum())
        if count[k] >= MIN_PER_BIN:
            fc_freq[k]  = float(prob_fc[m].mean())
            obs_freq[k] = float(obs_binary[m].mean())
    return centers, fc_freq, obs_freq, count
 
 
events = ['p90', 'p10']
rel_centers     = np.linspace(0.05, 0.95, N_REL_BINS)
rel_fc_freq_arr = {ev: np.full((len(vars_verif), N_REL_BINS), np.nan) for ev in events}
rel_obs_freq_arr= {ev: np.full((len(vars_verif), N_REL_BINS), np.nan) for ev in events}
rel_count_arr   = {ev: np.zeros((len(vars_verif), N_REL_BINS), dtype=np.int64) for ev in events}
rel_base_rate   = {ev: np.full(len(vars_verif), np.nan) for ev in events}
 
for iv, var in enumerate(tqdm(vars_verif, desc='Reliability    ')):
    # Thresholds from ERA5 training-period climatology
    hist = ds_target_hist[var].values
    hist = hist[np.isfinite(hist)]
    if hist.size < 20:
        continue
    p90 = float(np.quantile(hist, 0.90))
    p10 = float(np.quantile(hist, 0.10))
 
    fc_ens, obs = gather_pairs(ds_dress_all[var], ds_target[var], has_member=True)
    if fc_ens.size == 0:
        continue
 
    for ev, thresh, exceed in [('p90', p90, True), ('p10', p10, False)]:
        # Forecast probability from ensemble frequency
        if exceed:
            prob_fc    = (fc_ens > thresh).mean(axis=1)
            obs_binary = (obs > thresh).astype(np.float64)
        else:
            prob_fc    = (fc_ens < thresh).mean(axis=1)
            obs_binary = (obs < thresh).astype(np.float64)
 
        c, ff, of, cnt = reliability_curve(prob_fc, obs_binary, N_REL_BINS)
        rel_fc_freq_arr[ev][iv, :]  = ff
        rel_obs_freq_arr[ev][iv, :] = of
        rel_count_arr[ev][iv, :]    = cnt
        rel_base_rate[ev][iv]       = float(obs_binary.mean()) if obs_binary.size else np.nan
 
ds_reli = xr.Dataset(
    {
        'fc_freq_p90':  (('variable', 'bin'), rel_fc_freq_arr['p90']),
        'obs_freq_p90': (('variable', 'bin'), rel_obs_freq_arr['p90']),
        'count_p90':    (('variable', 'bin'), rel_count_arr['p90']),
        'base_rate_p90':(('variable',),       rel_base_rate['p90']),
        'fc_freq_p10':  (('variable', 'bin'), rel_fc_freq_arr['p10']),
        'obs_freq_p10': (('variable', 'bin'), rel_obs_freq_arr['p10']),
        'count_p10':    (('variable', 'bin'), rel_count_arr['p10']),
        'base_rate_p10':(('variable',),       rel_base_rate['p10']),
    },
    coords={'variable': vars_verif, 'bin_center': ('bin', rel_centers)},
)
```

```python
# ============================================================================
# 2. Rank histogram (Talagrand)
# ============================================================================
def rank_histogram_counts(ens, obs, n_member):
    """For each (forecast, obs) pair, the rank of obs among the n_member+1
    sorted values (members + obs). Random tie-breaking handles duplicates."""
    n = obs.size
    if n == 0:
        return np.zeros(n_member + 1, dtype=np.int64)
    # Compare obs to each member; random tie-breaking by jittering ties
    rng = np.random.default_rng(0)
    jitter = rng.uniform(-1e-12, 1e-12, ens.shape)
    ranks = (ens + jitter < obs[:, None]).sum(axis=1)        # 0 .. n_member
    counts = np.bincount(ranks, minlength=n_member + 1)
    return counts
 
 
n_member = ds_dress_all.sizes['member']
N_RANK_BINS = n_member + 1
 
rank_counts = np.zeros((len(vars_verif), N_RANK_BINS), dtype=np.int64)
rank_total  = np.zeros(len(vars_verif), dtype=np.int64)
 
for iv, var in enumerate(tqdm(vars_verif, desc='Rank histogram ')):
    fc_ens, obs = gather_pairs(ds_dress_all[var], ds_target[var], has_member=True)
    if fc_ens.size == 0:
        continue
    rank_counts[iv, :] = rank_histogram_counts(fc_ens, obs, n_member)
    rank_total[iv]     = int(obs.size)
 
# Expected frequency under perfect calibration: total / (n_member + 1)
rank_expected = rank_total / N_RANK_BINS
rank_freq     = np.where(rank_total[:, None] > 0,
                         rank_counts / np.maximum(rank_total[:, None], 1),
                         np.nan)
 
ds_rank = xr.Dataset(
    {
        'count':    (('variable', 'rank'), rank_counts),
        'freq':     (('variable', 'rank'), rank_freq),
        'n_pairs':  (('variable',), rank_total),
        'expected_freq': (('variable',), 1.0 / N_RANK_BINS * np.ones(len(vars_verif))),
    },
    coords={'variable': vars_verif, 'rank': np.arange(N_RANK_BINS)},
)
```

```python
# ============================================================================
# 3. Spread-skill relationship
# ============================================================================
def spread_skill_binned(spread, error, n_bins=N_SPREAD_BINS):
    """Bin by ensemble spread, then within each bin compute the RMSE of the
    ensemble-mean error. For a calibrated ensemble, mean spread ~ RMSE per bin
    (under the convention that spread = sqrt((1 + 1/M) * var_ensemble),
    or approximately spread = ensemble std for M >> 1)."""
    valid = np.isfinite(spread) & np.isfinite(error)
    spread, error = spread[valid], error[valid]
    if spread.size < n_bins * MIN_PER_BIN:
        return (np.full(n_bins, np.nan),) * 4
 
    # Equal-count bins (quantile edges) -- robust to skewed spread distributions
    edges = np.quantile(spread, np.linspace(0.0, 1.0, n_bins + 1))
    edges[0] -= 1e-12   # guard against the lower edge missing the min sample
    idx = np.clip(np.digitize(spread, edges) - 1, 0, n_bins - 1)
 
    mean_spread = np.full(n_bins, np.nan)
    rmse_err    = np.full(n_bins, np.nan)
    count       = np.zeros(n_bins, dtype=np.int64)
    centers     = np.full(n_bins, np.nan)
 
    for k in range(n_bins):
        m = (idx == k)
        count[k] = int(m.sum())
        if count[k] >= MIN_PER_BIN:
            mean_spread[k] = float(spread[m].mean())
            rmse_err[k]    = float(np.sqrt(np.mean(error[m] ** 2)))
            centers[k]     = 0.5 * (edges[k] + edges[k + 1])
    return mean_spread, rmse_err, count, centers
 
 
spread_arr  = np.full((len(vars_verif), N_SPREAD_BINS), np.nan)
rmse_arr    = np.full((len(vars_verif), N_SPREAD_BINS), np.nan)
count_arr   = np.zeros((len(vars_verif), N_SPREAD_BINS), dtype=np.int64)
center_arr  = np.full((len(vars_verif), N_SPREAD_BINS), np.nan)
ratio_arr   = np.full(len(vars_verif), np.nan)  # overall mean_spread / rmse
 
for iv, var in enumerate(tqdm(vars_verif, desc='Spread-skill   ')):
    fc_ens, obs = gather_pairs(ds_dress_all[var], ds_target[var], has_member=True)
    if fc_ens.size == 0:
        continue
 
    fc_mean = fc_ens.mean(axis=1)
    # Use sample std (ddof=1) so spread is interpretable as the per-forecast
    # predictive standard deviation; for M=50 the ddof correction is ~1%.
    fc_spread = fc_ens.std(axis=1, ddof=1)
    err = fc_mean - obs
 
    s, r, c, ctr = spread_skill_binned(fc_spread, err, N_SPREAD_BINS)
    spread_arr[iv, :] = s
    rmse_arr[iv, :]   = r
    count_arr[iv, :]  = c
    center_arr[iv, :] = ctr
 
    # Aggregate calibration ratio: 1.0 == perfectly calibrated spread
    overall_spread = float(np.sqrt(np.mean(fc_spread ** 2)))
    overall_rmse   = float(np.sqrt(np.mean(err ** 2)))
    ratio_arr[iv]  = overall_spread / overall_rmse if overall_rmse > 0 else np.nan
 
ds_spread = xr.Dataset(
    {
        'mean_spread':   (('variable', 'bin'), spread_arr),
        'rmse_error':    (('variable', 'bin'), rmse_arr),
        'bin_center':    (('variable', 'bin'), center_arr),
        'count':         (('variable', 'bin'), count_arr),
        'spread_rmse_ratio': (('variable',),   ratio_arr),
    },
    coords={'variable': vars_verif, 'bin': np.arange(N_SPREAD_BINS)},
)
```

```python
# ============================================================================
# Plot config (tweak these to re-bin without re-running verification)
# ============================================================================
N_REL_BINS_PLOT    = N_REL_BINS    # must match N_REL_BINS used to build ds_reli
N_RANK_BINS_PLOT   = 51    # must match n_member + 1
N_SPREAD_BINS_PLOT = N_SPREAD_BINS    # must match N_SPREAD_BINS used to build ds_spread

N_COLS = 4                 # panel grid columns
FIGSIZE_PER_PANEL = (3.0, 3.0)


def _grid(n_vars, n_cols=N_COLS, panel_size=FIGSIZE_PER_PANEL):
    n_rows = int(np.ceil(n_vars / n_cols))
    fig, axes = plt.subplots(
        n_rows, n_cols,
        figsize=(panel_size[0] * n_cols, panel_size[1] * n_rows),
        squeeze=False,
    )
    return fig, axes.flat, n_rows

# ============================================================================
# 1. Reliability diagram
# ============================================================================
def plot_reliability(ds_reli, event='p90', title_prefix=''):
    """event in {'p90', 'p10'}."""
    vars_list = ds_reli['variable'].values
    fig, axflat, _ = _grid(len(vars_list))

    fc_freq  = ds_reli[f'fc_freq_{event}'].values     # (n_var, n_bin)
    obs_freq = ds_reli[f'obs_freq_{event}'].values
    count    = ds_reli[f'count_{event}'].values
    base     = ds_reli[f'base_rate_{event}'].values

    for iv, var in enumerate(vars_list):
        ax = axflat[iv]
        ax.plot([0, 1], [0, 1], 'k--', lw=0.8, alpha=0.5)        # perfect-reliability diagonal
        if np.isfinite(base[iv]):
            ax.axhline(base[iv], color='gray', lw=0.6, ls=':')   # climatology line
            ax.axvline(base[iv], color='gray', lw=0.6, ls=':')

        ff = fc_freq[iv]
        of = obs_freq[iv]
        cnt = count[iv]
        m = np.isfinite(ff) & np.isfinite(of)
        if m.any():
            # Marker size scales with bin count
            sizes = 8 + 60 * np.sqrt(cnt[m] / max(cnt[m].max(), 1))
            ax.scatter(ff[m], of[m], s=sizes, c='C0', alpha=0.8, edgecolor='k', lw=0.4)
            ax.plot(ff[m], of[m], '-', color='C0', lw=1.0, alpha=0.5)

        ax.set_xlim(0, 1); ax.set_ylim(0, 1)
        ax.set_aspect('equal', adjustable='box')
        ax.set_title(var, fontsize=8)
        ax.tick_params(labelsize=7)
        if iv % N_COLS == 0:
            ax.set_ylabel('Observed freq.', fontsize=8)
        if iv >= len(vars_list) - N_COLS:
            ax.set_xlabel('Forecast prob.', fontsize=8)

    for k in range(len(vars_list), len(list(axflat))):
        axflat[k].axis('off')

    fig.suptitle(f'{title_prefix}Reliability diagram ({event})', fontsize=10, y=1.00)
    fig.tight_layout()
    return fig


# Calls (run one per event)
fig_rel_p90 = plot_reliability(ds_reli, event='p90')
fig_rel_p10 = plot_reliability(ds_reli, event='p10')
```

```python
# ============================================================================
# 2. Rank histogram
# ============================================================================
def plot_rank_histogram(ds_rank, n_aggregate=1):
    """n_aggregate>1 collapses adjacent rank bins for a smoother plot.
    Useful when n_member+1 = 51 is too noisy at the per-rank level."""
    vars_list = ds_rank['variable'].values
    fig, axflat, _ = _grid(len(vars_list))

    counts   = ds_rank['count'].values                 # (n_var, n_rank)
    n_pairs  = ds_rank['n_pairs'].values
    n_rank   = counts.shape[1]
    n_groups = n_rank // n_aggregate

    for iv, var in enumerate(vars_list):
        ax = axflat[iv]
        if n_pairs[iv] == 0:
            ax.set_visible(False)
            continue

        c = counts[iv, :n_groups * n_aggregate].reshape(n_groups, n_aggregate).sum(axis=1)
        freq = c / n_pairs[iv]
        expected = n_aggregate / n_rank                # uniform reference

        edges = np.linspace(0, n_rank, n_groups + 1)
        centers = 0.5 * (edges[:-1] + edges[1:])
        width = edges[1] - edges[0]

        ax.bar(centers, freq, width=width * 0.95,
               color='C0', alpha=0.7, edgecolor='k', lw=0.3)
        ax.axhline(expected, color='red', lw=0.8, ls='--')

        ax.set_xlim(0, n_rank)
        ax.set_ylim(0, max(freq.max() * 1.2, expected * 2))
        ax.set_title(var, fontsize=8)
        ax.tick_params(labelsize=7)
        if iv % N_COLS == 0:
            ax.set_ylabel('Frequency', fontsize=8)
        if iv >= len(vars_list) - N_COLS:
            ax.set_xlabel('Rank of obs in ensemble', fontsize=8)

    for k in range(len(vars_list), len(list(axflat))):
        axflat[k].axis('off')

    fig.suptitle(f'Rank histogram (n_aggregate={n_aggregate})', fontsize=10, y=1.00)
    fig.tight_layout()
    return fig


# Calls (try n_aggregate=1 for full resolution, 5 or 10 for smoother bars)
fig_rank_fine   = plot_rank_histogram(ds_rank, n_aggregate=1)
fig_rank_coarse = plot_rank_histogram(ds_rank, n_aggregate=5)
```

```python
# ============================================================================
# 3. Spread-skill relationship
# ============================================================================
def plot_spread_skill(ds_spread):
    vars_list = ds_spread['variable'].values
    fig, axflat, _ = _grid(len(vars_list))

    mean_spread = ds_spread['mean_spread'].values
    rmse_err    = ds_spread['rmse_error'].values
    ratio       = ds_spread['spread_rmse_ratio'].values

    for iv, var in enumerate(vars_list):
        ax = axflat[iv]
        s = mean_spread[iv]
        r = rmse_err[iv]
        m = np.isfinite(s) & np.isfinite(r)
        if not m.any():
            ax.set_visible(False)
            continue

        # 1:1 reference line spanning the data range
        lo = float(min(s[m].min(), r[m].min()))
        hi = float(max(s[m].max(), r[m].max()))
        pad = 0.05 * (hi - lo)
        ax.plot([lo - pad, hi + pad], [lo - pad, hi + pad],
                'k--', lw=0.8, alpha=0.5)

        ax.plot(s[m], r[m], 'o-', color='C0', lw=1.0, ms=4,
                markeredgecolor='k', markeredgewidth=0.4)

        ax.set_xlim(lo - pad, hi + pad)
        ax.set_ylim(lo - pad, hi + pad)
        ax.set_aspect('equal', adjustable='box')
        ax.set_title(f'{var}\nratio={ratio[iv]:.2f}', fontsize=7)
        ax.tick_params(labelsize=7)
        if iv % N_COLS == 0:
            ax.set_ylabel('RMSE of mean', fontsize=8)
        if iv >= len(vars_list) - N_COLS:
            ax.set_xlabel('Mean spread', fontsize=8)

    for k in range(len(vars_list), len(list(axflat))):
        axflat[k].axis('off')

    fig.suptitle('Spread–skill relationship (1:1 = perfect calibration)',
                 fontsize=10, y=1.00)
    fig.tight_layout()
    return fig


fig_ss = plot_spread_skill(ds_spread)
```
