````python
import os
import numpy as np
import pandas as pd
import xarray as xr
from tqdm import tqdm
 
from scipy.stats import norm
from scipy.optimize import minimize
````

````python
# ----------------------------- config ---------------------------------------
EPS = 1e-6
MIN_SAMPLES = 10
N_DRESS = 50
RNG_SEED = 42


years_train = np.arange(1959, 2000)
years_verif = np.arange(2000, 2021)
T_CENTER = float(years_train.mean())   # e.g., 1979.0 for 1959..1999

# all Gaussian-distributed variables to post-process
vars_keep = [
    'Pituffik_FT_days',
    'Fairbanks_T2_1d_max', 'Fairbanks_T2_1h_max', 'Fairbanks_T2_30d_max', 'Fairbanks_T2_7d_max', 
    'Fort_Bragg_SPEI_09_mean', 'Fort_Bragg_SPEI_09_min', 'Fort_Bragg_SPEI_48_mean', 'Fort_Bragg_SPEI_48_min',
    'Guam_SPEI_03_mean', 'Guam_SPEI_03_min', 'Guam_SPEI_48_mean', 'Guam_SPEI_48_min',
    'Yuma_PG_SPEI_24_mean', 'Yuma_PG_SPEI_24_min', 'Yuma_PG_SPEI_48_mean', 'Yuma_PG_SPEI_48_min'
]

data_dir = '/glade/derecho/scratch/ksha/EPRI_data/METRICS/'
in_input  = os.path.join(data_dir, 'STN_CESM_ALL.zarr')
in_target = os.path.join(data_dir, 'STN_ERA5_ALL.zarr')

# ----------------------------- load -----------------------------------------
ds_input = xr.open_zarr(in_input)[vars_keep].drop_vars('valid_year', errors='ignore')
ds_input_train = ds_input.sel(init_year=years_train)
ds_input_verif = ds_input.sel(init_year=years_verif)
 
ds_target = xr.open_zarr(in_target)[vars_keep]
 
n_member = ds_input_train.sizes['member']
n_lead   = ds_input_train.sizes['lead_year']
n_train  = ds_input_train.sizes['init_year']
n_verif  = ds_input_verif.sizes['init_year']
 
lead_years        = ds_input_train['lead_year'].values
init_years_train  = ds_input_train['init_year'].values
init_years_verif  = ds_input_verif['init_year'].values
````

````python
# ============================================================================
# Helpers
# ============================================================================

def fit_bma_gaussian_lead(x_train, y_train, max_iter=200, tol=1e-6):
    """
    Exchangeable Gaussian BMA for one (variable, lead_year):
        p(y|x) = (1/K) sum_k N(y | a + b x_k, sigma^2)
    (a, b) by OLS of y on ensemble mean; sigma^2 by EM (Fraley et al. 2010)
    with equal weights and (a, b) fixed.

    x_train : (T, K)   y_train : (T,)
    """
    x_bar = x_train.mean(axis=1)
    mask = (np.isfinite(x_bar) & np.isfinite(y_train)
            & np.all(np.isfinite(x_train), axis=1))
    if mask.sum() < MIN_SAMPLES:
        return np.nan, np.nan, np.nan

    x, xb, y = x_train[mask], x_bar[mask], y_train[mask]
    n = len(y)

    # ---- OLS: y = a + b * x_bar ----------------------------------------
    xb_m, y_m = xb.mean(), y.mean()
    var_xb = np.sum((xb - xb_m) ** 2)
    if var_xb < EPS:
        a, b = y_m, 0.0
    else:
        b = np.sum((xb - xb_m) * (y - y_m)) / var_xb
        a = y_m - b * xb_m

    # ---- EM for sigma^2 (equal weights cancel in normalization) --------
    resid = y[:, None] - (a + b * x)            # (n, K)
    sigma2 = max(np.mean(resid ** 2), EPS)

    for _ in range(max_iter):
        s2_old = sigma2
        log_p = -0.5 * resid ** 2 / sigma2
        log_p -= log_p.max(axis=1, keepdims=True)   # log-sum-exp stabilization
        z = np.exp(log_p)
        z /= z.sum(axis=1, keepdims=True)           # E-step: posterior weights
        sigma2 = max(np.sum(z * resid ** 2) / n, EPS)  # M-step
        if abs(sigma2 - s2_old) / s2_old < tol:
            break
    return a, b, sigma2


def crps_ensemble(ens, y):
    """Sample-based CRPS. ens: (..., M), y: (...,). Biased 1/M^2 estimator."""
    y_exp = y[..., None]
    term1 = np.nanmean(np.abs(ens - y_exp), axis=-1)
    diff = ens[..., :, None] - ens[..., None, :]
    term2 = 0.5 * np.nanmean(np.abs(diff), axis=(-2, -1))
    return term1 - term2


def acc_pearson(f, y):
    m = np.isfinite(f) & np.isfinite(y)
    if m.sum() < 3:
        return np.nan
    return np.corrcoef(f[m], y[m])[0, 1]

def lookup_target(y_full, valid_years_needed, valid_years_avail):
    out = np.full(len(valid_years_needed), np.nan, dtype=np.float64)
    in_mask = np.isin(valid_years_needed, valid_years_avail)
    if in_mask.any():
        out[in_mask] = y_full.sel(valid_year=valid_years_needed[in_mask]).values
    return out
````

````python
# ============================================================================
# 1. Gaussian BMA fitting
# ============================================================================

ds_params = xr.Dataset(coords={'lead_year': lead_years})

for var in vars_keep:
    a_vec  = np.full(n_lead, np.nan)
    b_vec  = np.full(n_lead, np.nan)
    s2_vec = np.full(n_lead, np.nan)

    x_full = ds_input_train[var].values        # (init_year, lead_year, member)

    for li, ly in enumerate(lead_years):
        valid_years_t = init_years_train + int(ly)
        y_t = (ds_target[var]
               .reindex(valid_year=valid_years_t).values)   # NaN-fill if missing
        x_t = x_full[:, li, :]
        a_vec[li], b_vec[li], s2_vec[li] = fit_bma_gaussian_lead(x_t, y_t)

    ds_params[f'{var}_a']      = ('lead_year', a_vec)
    ds_params[f'{var}_b']      = ('lead_year', b_vec)
    ds_params[f'{var}_sigma2'] = ('lead_year', s2_vec)
````

````python
# ============================================================================
# 2. Calibration: BMA predictive mean & std on verification ensemble
#       mu    = a + b * x_bar
#       var   = sigma^2 + b^2 * (1/K) sum_k (x_k - x_bar)^2
# ============================================================================

ds_calib = xr.Dataset(
    coords={'init_year': init_years_verif, 'lead_year': lead_years}
)

for var in vars_keep:
    x_v = ds_input_verif[var]                       # (init_year, lead_year, member)
    a   = ds_params[f'{var}_a']
    b   = ds_params[f'{var}_b']
    s2  = ds_params[f'{var}_sigma2']

    x_bar = x_v.mean(dim='member')
    s_x2  = ((x_v - x_bar) ** 2).mean(dim='member')  # population variance /K

    mu_bma  = a + b * x_bar
    var_bma = (s2 + (b ** 2) * s_x2).clip(min=EPS)
    std_bma = np.sqrt(var_bma)

    ds_calib[f'{var}_mean'] = mu_bma.transpose('init_year', 'lead_year')
    ds_calib[f'{var}_std']  = std_bma.transpose('init_year', 'lead_year')
````

````python
# ============================================================================
# 3. Ensemble dressing: 50 samples ~ N(mu_BMA, sigma_BMA^2)
# ============================================================================

rng = np.random.default_rng(RNG_SEED)

ds_dress = xr.Dataset(
    coords={
        'init_year':       init_years_verif,
        'lead_year':       lead_years,
        'dressed_member':  np.arange(N_DRESS),
    }
)

for var in vars_keep:
    mu = ds_calib[f'{var}_mean'].values   # (n_verif, n_lead)
    sd = ds_calib[f'{var}_std'].values
    z  = rng.standard_normal((n_verif, n_lead, N_DRESS))
    ds_dress[var] = (
        ('init_year', 'lead_year', 'dressed_member'),
        mu[..., None] + sd[..., None] * z,
    )
````

````python
# ============================================================================
# Step 4: Verification (CRPS and ACC)
# ============================================================================
def crps_gaussian(mu, sigma, y):
    """Closed-form CRPS for N(mu, sigma^2)."""
    sigma = np.maximum(sigma, EPS)
    z = (y - mu) / sigma
    return sigma * (z * (2.0 * norm.cdf(z) - 1.0)
                    + 2.0 * norm.pdf(z)
                    - 1.0 / np.sqrt(np.pi))

def crps_ensemble(ens, y):
    out = np.full(ens.shape[0], np.nan)
    finite_ens = np.all(np.isfinite(ens), axis=1)
    mask = finite_ens & np.isfinite(y)
    if not mask.any():
        return out
    e = ens[mask]
    yv = y[mask]
    term1 = np.mean(np.abs(e - yv[:, None]), axis=1)
    diff = np.abs(e[:, :, None] - e[:, None, :])
    term2 = 0.5 * diff.mean(axis=(1, 2))
    out[mask] = term1 - term2
    return out
 
# ---- (b) per-system climatology --------------------------------------------
def anomaly_correlation(fc, obs, fc_clim=None, obs_clim=None):
    """ACC with optionally separate forecast and observation climatologies.
    If clims are None, each is taken as the mean of its own series over the
    (finite) sample -- equivalent to Pearson r."""
    m = np.isfinite(fc) & np.isfinite(obs)
    if m.sum() < 2:
        return np.nan
    fc_m, obs_m = fc[m], obs[m]
    if fc_clim is None:
        fc_clim = float(np.mean(fc_m))
    if obs_clim is None:
        obs_clim = float(np.mean(obs_m))
    fa = fc_m  - fc_clim
    oa = obs_m - obs_clim
    den = np.sqrt(np.sum(fa * fa) * np.sum(oa * oa))
    return float(np.sum(fa * oa) / den) if den > 0 else np.nan
 
verif_rows = []
for var in tqdm(vars_keep, desc='Verification   '):
    y_full = ds_target[var]
    valid_years_avail = y_full['valid_year'].values
 
    x_raw   = ds_input_verif[var].values
    x_dress = ds_dress[var].values
    mu      = ds_calib[var + '_mean'].values
    sigma   = ds_calib[var + '_std'].values
 
    for il, lead in enumerate(lead_years):
        valid_yrs = init_years_verif + int(lead)
        y_arr = lookup_target(y_full, valid_yrs, valid_years_avail)
 
        # CRPS
        crps_raw    = crps_ensemble(x_raw[:, il, :],   y_arr)
        crps_dress  = crps_ensemble(x_dress[:, il, :], y_arr)
        crps_closed = crps_gaussian(mu[:, il], sigma[:, il], y_arr)
 
        # ACC with per-system climatology (each centered on its own mean)
        raw_mean   = x_raw[:, il, :].mean(axis=1)
        dress_mean = x_dress[:, il, :].mean(axis=1)
        calib_mean = mu[:, il]
 
        acc_raw   = anomaly_correlation(raw_mean,   y_arr)
        acc_dress = anomaly_correlation(dress_mean, y_arr)
        acc_calib = anomaly_correlation(calib_mean, y_arr)
 
        # Per-system clims (recorded for transparency)
        fin_y      = np.isfinite(y_arr)
        clim_obs   = float(np.nanmean(y_arr))      if fin_y.any() else np.nan
        clim_raw   = float(np.nanmean(raw_mean))   if np.isfinite(raw_mean).any()   else np.nan
        clim_calib = float(np.nanmean(calib_mean)) if np.isfinite(calib_mean).any() else np.nan
 
        mean_raw    = float(np.nanmean(crps_raw))
        mean_dress  = float(np.nanmean(crps_dress))
        mean_closed = float(np.nanmean(crps_closed))
        crpss = (1.0 - mean_closed / mean_raw) if (mean_raw and mean_raw > 0) else np.nan
 
        verif_rows.append({
            'variable':    var,
            'lead_year':   int(lead),
            'n_eff':       int(fin_y.sum()),
            'clim_obs':    clim_obs,
            'clim_raw':    clim_raw,
            'clim_calib':  clim_calib,
            'CRPS_raw':    mean_raw,
            'CRPS_dress':  mean_dress,
            'CRPS_closed': mean_closed,
            'CRPSS':       crpss,
            'ACC_raw':     acc_raw,
            'ACC_dress':   acc_dress,
            'ACC_calib':   acc_calib,
        })
 
df_verif = pd.DataFrame(verif_rows)
````

````python
save_dir = '/glade/derecho/scratch/ksha/EPRI_data/PP_calib/BMA/'
ds_dress.to_zarr(save_dir+'Dress.zarr', mode='w')
ds_calib.to_zarr(save_dir+'Calib.zarr', mode='w')
````
