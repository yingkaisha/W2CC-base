````python
import os
import numpy as np
import pandas as pd
import xarray as xr
from tqdm import tqdm
 
from scipy.stats import norm
from scipy.optimize import minimize
from scipy.interpolate import interp1d
````

````python
# ----------------------------- config ---------------------------------------
EPS = 1e-6
MIN_SAMPLES = 10
N_DRESS = 50
RNG_SEED = 42
N_QUANTILES = 200

q_levels = np.linspace(0.0025, 0.9975, N_QUANTILES)
rng_qt    = np.random.default_rng(RNG_SEED)
rng_dress = np.random.default_rng(RNG_SEED + 1)


years_train = np.arange(1959, 2000)
years_verif = np.arange(2000, 2021)
T_CENTER = float(years_train.mean())   # e.g., 1979.0 for 1959..1999

# non-Gaussian variables to post-process
vars_precip = vars_keep = [
    'Fort_Bragg_PRECT_1d_max',
    'Fort_Bragg_PRECT_3d_max',
    'Fort_Bragg_PRECT_5d_max',
    'Pituffik_MDD'
]

data_dir = '/glade/derecho/scratch/ksha/EPRI_data/METRICS/'
in_input  = os.path.join(data_dir, 'STN_CESM_ALL.zarr')
in_target = os.path.join(data_dir, 'STN_ERA5_ALL.zarr')

# ----------------------------- load -----------------------------------------
ds_input = xr.open_zarr(in_input)[vars_keep].drop_vars('valid_year', errors='ignore')
ds_input_train = ds_input.sel(init_year=years_train)
ds_input_verif = ds_input.sel(init_year=years_verif)
 
ds_target = xr.open_zarr(in_target)[vars_keep]
 
n_member = ds_input_train.sizes['member']
n_lead   = ds_input_train.sizes['lead_year']
n_train  = ds_input_train.sizes['init_year']
n_verif  = ds_input_verif.sizes['init_year']
 
lead_years        = ds_input_train['lead_year'].values
init_years_train  = ds_input_train['init_year'].values
init_years_verif  = ds_input_verif['init_year'].values
````

````python
# ============================================================================
# Normal quantile transform (user-supplied, verbatim)
# ============================================================================
def build_quantile_table(samples):
    """Build empirical quantile table from 1D samples."""
    valid = samples[~np.isnan(samples)]
    if len(valid) < MIN_SAMPLES:
        return np.full(N_QUANTILES, np.nan), np.nan
    q_values = np.quantile(valid, q_levels).astype(np.float32)
    wet_frac = np.float32(np.mean(valid > 0.0))
    return q_values, wet_frac
 
 
def extrapolate_upper_tail(q_vals, q_levs):
    """Linearly extrapolate to quantile level 1.0."""
    v1, v2 = q_vals[-2], q_vals[-1]
    l1, l2 = q_levs[-2], q_levs[-1]
    slope = (v2 - v1) / max(l2 - l1, EPS)
    v_max = max(v2 + slope * (1.0 - l2), v2)
    return np.append(q_vals, np.float32(v_max)), np.append(q_levs, 1.0)
 
 
def precip_to_gaussian(values, q_values, wet_frac, rng=None):
    """Map precipitation -> Gaussian via NQT."""
    n = len(values)
    gaussian = np.full(n, np.nan, dtype=np.float32)
 
    if np.all(np.isnan(q_values)):
        return gaussian
    if rng is None:
        rng = np.random.default_rng()
 
    valid = ~np.isnan(values)
    vals = np.maximum(values[valid], 0.0)
 
    dry_frac = 1.0 - wet_frac
    u = np.full(len(vals), np.nan, dtype=np.float64)
 
    is_zero = vals <= 0.0
    is_wet = ~is_zero
 
    if dry_frac > 0 and is_zero.any():
        u[is_zero] = rng.uniform(EPS, max(dry_frac, 2 * EPS), size=is_zero.sum())
 
    if is_wet.any():
        wet_q_mask = q_values > 0.0
        if wet_q_mask.sum() >= 2:
            q_wet = q_values[wet_q_mask]
            l_wet = q_levels[wet_q_mask]
            q_ext, l_ext = extrapolate_upper_tail(q_wet, l_wet)
 
            q_unique, idx_unique = np.unique(q_ext, return_index=True)
            ql_unique = l_ext[idx_unique]
 
            if len(q_unique) >= 2:
                interp_func = interp1d(
                    q_unique, ql_unique,
                    kind='linear', bounds_error=False,
                    fill_value=(ql_unique[0], ql_unique[-1]),
                )
                u_wet_raw = interp_func(vals[is_wet])
                l_min, l_max = ql_unique[0], ql_unique[-1]
                u[is_wet] = dry_frac + (1.0 - dry_frac) * (
                    (u_wet_raw - l_min) / max(l_max - l_min, EPS)
                )
            else:
                u[is_wet] = dry_frac + 0.5 * (1.0 - dry_frac)
        else:
            u[is_wet] = dry_frac + 0.5 * (1.0 - dry_frac)
 
    u = np.clip(u, 1e-6, 1.0 - 1e-6)
    gaussian[valid] = norm.ppf(u).astype(np.float32)
    return gaussian
 
 
def gaussian_to_precip(gaussian_values, q_values, wet_frac):
    """Inverse NQT: Gaussian -> precipitation (mm/day)."""
    shape = gaussian_values.shape
    g_flat = gaussian_values.ravel()
    precip = np.full_like(g_flat, np.nan, dtype=np.float32)
 
    if np.all(np.isnan(q_values)):
        return precip.reshape(shape)
 
    valid = ~np.isnan(g_flat)
    if not valid.any():
        return precip.reshape(shape)
 
    u = norm.cdf(g_flat[valid])
    dry_frac = 1.0 - wet_frac
    result = np.zeros_like(u, dtype=np.float32)
 
    wet_mask = u > dry_frac
    if wet_mask.any():
        wet_q_mask = q_values > 0.0
        if wet_q_mask.sum() >= 2:
            q_wet = q_values[wet_q_mask]
            l_wet = q_levels[wet_q_mask]
            q_ext, l_ext = extrapolate_upper_tail(q_wet, l_wet)
 
            q_unique, idx_unique = np.unique(q_ext, return_index=True)
            ql_unique = l_ext[idx_unique]
 
            if len(q_unique) >= 2:
                l_min, l_max = ql_unique[0], ql_unique[-1]
                u_rescaled = l_min + (l_max - l_min) * (
                    (u[wet_mask] - dry_frac) / max(1.0 - dry_frac, EPS)
                )
                inv_func = interp1d(
                    ql_unique, q_unique,
                    kind='linear', bounds_error=False,
                    fill_value=(q_unique[0], q_unique[-1]),
                )
                result[wet_mask] = np.maximum(inv_func(u_rescaled), 0.0)
            else:
                result[wet_mask] = q_values[wet_q_mask].mean()
        elif wet_q_mask.any():
            result[wet_mask] = q_values[wet_q_mask][0]
 
    precip[valid] = result.astype(np.float32)
    return precip.reshape(shape)
````

````python
# ============================================================================
# Helpers
# ============================================================================

def fit_bma_gaussian_lead(x_train, y_train, max_iter=200, tol=1e-6):
    """
    Exchangeable Gaussian BMA for one (variable, lead_year):
        p(y|x) = (1/K) sum_k N(y | a + b x_k, sigma^2)
    (a, b) by OLS of y on ensemble mean; sigma^2 by EM (Fraley et al. 2010)
    with equal weights and (a, b) fixed.

    x_train : (T, K)   y_train : (T,)
    """
    x_bar = x_train.mean(axis=1)
    mask = (np.isfinite(x_bar) & np.isfinite(y_train)
            & np.all(np.isfinite(x_train), axis=1))
    if mask.sum() < MIN_SAMPLES:
        return np.nan, np.nan, np.nan

    x, xb, y = x_train[mask], x_bar[mask], y_train[mask]
    n = len(y)

    # ---- OLS: y = a + b * x_bar ----------------------------------------
    xb_m, y_m = xb.mean(), y.mean()
    var_xb = np.sum((xb - xb_m) ** 2)
    if var_xb < EPS:
        a, b = y_m, 0.0
    else:
        b = np.sum((xb - xb_m) * (y - y_m)) / var_xb
        a = y_m - b * xb_m

    # ---- EM for sigma^2 (equal weights cancel in normalization) --------
    resid = y[:, None] - (a + b * x)            # (n, K)
    sigma2 = max(np.mean(resid ** 2), EPS)

    for _ in range(max_iter):
        s2_old = sigma2
        log_p = -0.5 * resid ** 2 / sigma2
        log_p -= log_p.max(axis=1, keepdims=True)   # log-sum-exp stabilization
        z = np.exp(log_p)
        z /= z.sum(axis=1, keepdims=True)           # E-step: posterior weights
        sigma2 = max(np.sum(z * resid ** 2) / n, EPS)  # M-step
        if abs(sigma2 - s2_old) / s2_old < tol:
            break
    return a, b, sigma2


def crps_ensemble(ens, y):
    """Sample-based CRPS. ens: (..., M), y: (...,). Biased 1/M^2 estimator."""
    y_exp = y[..., None]
    term1 = np.nanmean(np.abs(ens - y_exp), axis=-1)
    diff = ens[..., :, None] - ens[..., None, :]
    term2 = 0.5 * np.nanmean(np.abs(diff), axis=(-2, -1))
    return term1 - term2


def acc_pearson(f, y):
    m = np.isfinite(f) & np.isfinite(y)
    if m.sum() < 3:
        return np.nan
    return np.corrcoef(f[m], y[m])[0, 1]

def lookup_target(y_full, valid_years_needed, valid_years_avail):
    out = np.full(len(valid_years_needed), np.nan, dtype=np.float64)
    in_mask = np.isin(valid_years_needed, valid_years_avail)
    if in_mask.any():
        out[in_mask] = y_full.sel(valid_year=valid_years_needed[in_mask]).values
    return out
````

````python
# ----------------------------------------------------------------------------
# 1. Build quantile tables from training data
#       - ERA5: per variable, pooled over training valid_years
#       - SMYLE: per (variable, lead_year), pooled over init_year × member
# ----------------------------------------------------------------------------

qt_era5  = {}                  # var       -> (q_values, wet_frac)
qt_smyle = {}                  # (var, li) -> (q_values, wet_frac)

for var in vars_precip:
    y_train_all = (ds_target[var]
                   .sel(valid_year=slice(int(years_train.min()),
                                         int(years_train.max())))
                   .values.ravel())
    qt_era5[var] = build_quantile_table(y_train_all)

    x_full = ds_input_train[var].values            # (init, lead, member)
    for li in range(n_lead):
        qt_smyle[(var, li)] = build_quantile_table(x_full[:, li, :].ravel())


def transform_members(x_arr, q_values, wet_frac, rng):
    """Apply precip_to_gaussian member-wise. x_arr: (T, K) -> (T, K)."""
    out = np.full_like(x_arr, np.nan, dtype=np.float32)
    for k in range(x_arr.shape[1]):
        out[:, k] = precip_to_gaussian(x_arr[:, k], q_values, wet_frac, rng=rng)
    return out
````

````python
# ----------------------------------------------------------------------------
# 2. Fit BMA in Gaussian space  (reuses fit_bma_gaussian_lead)
# ----------------------------------------------------------------------------

ds_params_p = xr.Dataset(coords={'lead_year': lead_years})

for var in vars_precip:
    a_vec  = np.full(n_lead, np.nan)
    b_vec  = np.full(n_lead, np.nan)
    s2_vec = np.full(n_lead, np.nan)

    qe_v, wf_e = qt_era5[var]

    for li, ly in enumerate(lead_years):
        qs_v, wf_s = qt_smyle[(var, li)]
        if np.all(np.isnan(qe_v)) or np.all(np.isnan(qs_v)):
            continue

        valid_years_t = init_years_train + int(ly)
        y_t = (ds_target[var]
               .reindex(valid_year=valid_years_t).values)         # (T,)
        x_t = ds_input_train[var].isel(lead_year=li).values        # (T, K)

        y_g = precip_to_gaussian(y_t, qe_v, wf_e, rng=rng_qt)
        x_g = transform_members(x_t, qs_v, wf_s, rng_qt)

        a_vec[li], b_vec[li], s2_vec[li] = fit_bma_gaussian_lead(x_g, y_g)

    ds_params_p[f'{var}_a']      = ('lead_year', a_vec)
    ds_params_p[f'{var}_b']      = ('lead_year', b_vec)
    ds_params_p[f'{var}_sigma2'] = ('lead_year', s2_vec)
````

````python
# ----------------------------------------------------------------------------
# 3. Calibration: BMA mean & std in Gaussian space
#    (we do NOT store physical-space moments here — the back-transform is
#     non-linear, so means/variances must be computed from samples)
# ----------------------------------------------------------------------------

ds_calib_p = xr.Dataset(
    coords={'init_year': init_years_verif, 'lead_year': lead_years}
)

for var in vars_precip:
    a  = ds_params_p[f'{var}_a'].values
    b  = ds_params_p[f'{var}_b'].values
    s2 = ds_params_p[f'{var}_sigma2'].values

    mu_g  = np.full((n_verif, n_lead), np.nan, dtype=np.float32)
    sd_g  = np.full((n_verif, n_lead), np.nan, dtype=np.float32)

    x_full_v = ds_input_verif[var].values            # (init, lead, member)

    for li in range(n_lead):
        qs_v, wf_s = qt_smyle[(var, li)]
        if np.all(np.isnan(qs_v)) or not np.isfinite(s2[li]):
            continue

        x_g = transform_members(x_full_v[:, li, :], qs_v, wf_s, rng_qt)
        x_bar_g = np.nanmean(x_g, axis=1)
        s_x2_g  = np.nanmean((x_g - x_bar_g[:, None]) ** 2, axis=1)

        mu_g[:, li] = a[li] + b[li] * x_bar_g
        sd_g[:, li] = np.sqrt(np.maximum(s2[li] + (b[li] ** 2) * s_x2_g, EPS))

    ds_calib_p[f'{var}_mean_g'] = (('init_year', 'lead_year'), mu_g)
    ds_calib_p[f'{var}_std_g']  = (('init_year', 'lead_year'), sd_g)
````

````python
# ----------------------------------------------------------------------------
# 4. Dressing: 50 Gaussian samples, then back-transform via ERA5 inverse CDF
# ----------------------------------------------------------------------------

ds_dress_p = xr.Dataset(
    coords={
        'init_year':      init_years_verif,
        'lead_year':      lead_years,
        'dressed_member': np.arange(N_DRESS),
    }
)

for var in vars_precip:
    qe_v, wf_e = qt_era5[var]
    mu = ds_calib_p[f'{var}_mean_g'].values
    sd = ds_calib_p[f'{var}_std_g'].values

    z      = rng_dress.standard_normal((n_verif, n_lead, N_DRESS))
    samp_g = (mu[..., None] + sd[..., None] * z).astype(np.float32)
    samp_p = gaussian_to_precip(samp_g, qe_v, wf_e)         # shape preserved

    ds_dress_p[var] = (('init_year', 'lead_year', 'dressed_member'), samp_p)
````

````python
# ----------------------------------------------------------------------------
# 4b. Physical-space point forecasts: median (inv-NQT of BMA mean)
#     and mean (sample mean of dressed ensemble)
# ----------------------------------------------------------------------------

for var in vars_precip:
    qe_v, wf_e = qt_era5[var]

    # Predictive median: inverse-NQT of the Gaussian-space BMA mean.
    mu_g = ds_calib_p[f'{var}_mean_g'].values            # (n_verif, n_lead)
    median_p = gaussian_to_precip(mu_g.astype(np.float32), qe_v, wf_e)

    # Predictive mean: average dressed samples in physical space.
    mean_p = np.nanmean(ds_dress_p[var].values, axis=-1)  # (n_verif, n_lead)

    ds_calib_p[f'{var}_median'] = (('init_year', 'lead_year'), median_p)
    ds_calib_p[f'{var}_mean']   = (('init_year', 'lead_year'),
                                   mean_p.astype(np.float32))
````

````python
# ============================================================================
# 5. Verification (physical units)
# ============================================================================
def crps_ensemble(ens, y):
    out = np.full(ens.shape[0], np.nan)
    finite_ens = np.all(np.isfinite(ens), axis=1)
    mask = finite_ens & np.isfinite(y)
    if not mask.any():
        return out
    e = ens[mask]; yv = y[mask]
    term1 = np.mean(np.abs(e - yv[:, None]), axis=1)
    diff  = np.abs(e[:, :, None] - e[:, None, :])
    term2 = 0.5 * diff.mean(axis=(1, 2))
    out[mask] = term1 - term2
    return out
 
 
def anomaly_correlation(fc, obs):
    m = np.isfinite(fc) & np.isfinite(obs)
    if m.sum() < 2:
        return np.nan
    fa = fc[m]  - np.mean(fc[m])
    oa = obs[m] - np.mean(obs[m])
    den = np.sqrt(np.sum(fa * fa) * np.sum(oa * oa))
    return float(np.sum(fa * oa) / den) if den > 0 else np.nan
 
 
verif_rows = []
for var in tqdm(vars_keep, desc='Verification    '):
    y_full = ds_target[var]
    valid_years_avail = y_full['valid_year'].values
 
    x_raw   = ds_input_verif[var].values
    x_dress = ds_dress_p[var].values
 
    for il, lead in enumerate(lead_years):
        valid_yrs = init_years_verif + int(lead)
        y_arr = lookup_target(y_full, valid_yrs, valid_years_avail)
 
        crps_raw   = crps_ensemble(x_raw[:, il, :],   y_arr)
        crps_dress = crps_ensemble(x_dress[:, il, :], y_arr)
 
        raw_mean   = np.nanmean(x_raw[:, il, :],   axis=1)
        dress_mean = np.nanmean(x_dress[:, il, :], axis=1)
 
        acc_raw   = anomaly_correlation(raw_mean,   y_arr)
        acc_dress = anomaly_correlation(dress_mean, y_arr)
 
        mean_raw   = float(np.nanmean(crps_raw))
        mean_dress = float(np.nanmean(crps_dress))
        crpss = (1.0 - mean_dress / mean_raw) if (mean_raw and mean_raw > 0) else np.nan
        
        verif_rows.append({
            'variable':       var,
            'lead_year':      int(lead),
            'n_eff':          int(np.isfinite(y_arr).sum()),
            'CRPS_raw':       mean_raw,
            'CRPS_dress':     mean_dress,
            'CRPSS':          crpss,
            'ACC_raw':        acc_raw,
            'ACC_dress':      acc_dress,
        })
 
df_verif = pd.DataFrame(verif_rows)
````
