```python
years_verif = np.arange(2000, 2026)
years_train = np.arange(1968, 2000)

data_dir = '/glade/derecho/scratch/ksha/EPRI_data/METRICS/'
ds_target = xr.open_zarr(os.path.join(data_dir, 'STN_ERA5_ALL_20260604.zarr'))
ds_input  = xr.open_zarr(os.path.join(data_dir, 'STN_CESM_ALL_20260604.zarr'))

# ds_target_hist for determine percentiles
ds_target_hist = ds_target.sel(valid_year=years_train)

# raw forecast as reference
ds_raw_pred = ds_input.sel(init_year=np.arange(2000, 2021))

# ds_target with dims of `(valid_year: 26)`, valid_year = init_year + lead_year, up to 2025
ds_target = ds_target.sel(valid_year=years_verif)

# ====================================================================== #
# EMOS
# ====================================================================== #
save_dir = '/glade/derecho/scratch/ksha/EPRI_data/PP_calib/EMOS/'
ds_dress = xr.open_zarr(save_dir+'Dress_4param.zarr')
ds_dress_t = xr.open_zarr(save_dir+'Dress_Trend.zarr')
ds_dress_qm = xr.open_zarr(save_dir+'Dress_QM.zarr')

ds_calib = xr.open_zarr(save_dir+'Calib_4param.zarr')
ds_calib_t = xr.open_zarr(save_dir+'Calib_Trend.zarr')

ds_dress_qm_mean = ds_dress_qm.mean('member')
ds_dress_qm_mean = ds_dress_qm_mean.rename({v: f"{v}_mu" for v in ds_dress_qm_mean.data_vars})

ds_dress_qm_std = ds_dress_qm.std('member')
ds_dress_qm_std = ds_dress_qm_std.rename({v: f"{v}_sigma" for v in ds_dress_qm_std.data_vars})

ds_calib_qm = xr.merge([ds_dress_qm_mean, ds_dress_qm_std])

# ds_dress_all with dims of `(init_year: 21, lead_year: 10, member: 50)` for all post-processed variables
ds_dress_EMOS = xr.merge([ds_dress, ds_dress_t, ds_dress_qm])

# ds_calib_all with dims of `(init_year: 21, lead_year: 10)` for non-precip (Gaussian distributed) variables, each variable has suffix of `_mu` and `_sigma` for mean and std.
ds_calib_EMOS = xr.merge([ds_calib, ds_calib_t, ds_calib_qm])

# ====================================================================== #
# BMA
# ====================================================================== #
save_dir = '/glade/derecho/scratch/ksha/EPRI_data/PP_calib/BMA/'
ds_dress_bma = xr.open_zarr(save_dir+'Dress.zarr')
ds_calib_bma = xr.open_zarr(save_dir+'Calib.zarr')
ds_dress_qm_bma = xr.open_zarr(save_dir+'Dress_QM.zarr').rename({'dressed_member': 'member'})
ds_dress_bma = ds_dress_bma.rename({'dressed_member': 'member'})

qm_mu = ds_dress_qm_bma.mean('member').rename({v: f"{v}_mu" for v in ds_dress_qm_bma.data_vars})
qm_sg = ds_dress_qm_bma.std('member', ddof=1).rename({v: f"{v}_sigma" for v in ds_dress_qm_bma.data_vars})

ds_calib_BMA = xr.merge([ds_calib_bma, qm_mu, qm_sg])
ds_dress_BMA = xr.merge([ds_dress_bma, ds_dress_qm_bma])
```

```python
# ----------------------------------------------------------------------
# helpers: alignment
# ----------------------------------------------------------------------
 
def align_obs(ds_fcst, ds_target, var_obs, years_verif):
    """
    Build obs array on the forecast (init_year, lead_year) grid using
    valid_year = init_year + lead_year. Cells outside years_verif (or
    missing in ds_target) are NaN.
 
    Returns
    -------
    obs   : np.ndarray (n_init, n_lead)
    leads : np.ndarray of lead_year values
    """
    init = np.asarray(ds_fcst['init_year'].values)
    lead = np.asarray(ds_fcst['lead_year'].values)
    vy = init[:, None] + lead[None, :]
 
    tgt_years = np.asarray(ds_target['valid_year'].values)
    tgt_vals = np.asarray(ds_target[var_obs].values)
    lut = {int(y): v for y, v in zip(tgt_years, tgt_vals)}
 
    obs = np.full(vy.shape, np.nan)
    ok = np.isin(vy, years_verif)
    for i in range(vy.shape[0]):
        for l in range(vy.shape[1]):
            if ok[i, l] and int(vy[i, l]) in lut:
                obs[i, l] = lut[int(vy[i, l])]
    return obs, lead
 
 
def extract_source(var_, kind, ds, common_inits=None):
    """
    Returns dict with:
      mu     (init, lead)            ensemble mean / Gaussian mu
      sigma  (init, lead)            spread (member std, ddof=1 / Gaussian sigma)
      ens    (init, lead, member) or None
      ds     subset dataset (for coords)
    """
    if common_inits is not None:
        ds = ds.sel(init_year=ds['init_year'].isin(common_inits))
 
    if kind == 'gaussian':
        mu = ds[f'{var_}_mu'].transpose('init_year', 'lead_year').values
        sg = ds[f'{var_}_sigma'].transpose('init_year', 'lead_year').values
        sg = np.where(sg > 0, sg, np.nan)
        return dict(kind=kind, mu=mu, sigma=sg, ens=None, ds=ds)
    elif kind == 'ensemble':
        ens = ds[var_].transpose('init_year', 'lead_year', 'member').values
        mu = np.nanmean(ens, axis=-1)
        sg = np.nanstd(ens, axis=-1, ddof=1)
        return dict(kind=kind, mu=mu, sigma=sg, ens=ens, ds=ds)
    raise ValueError(f"unknown kind: {kind}")
 
 
# ----------------------------------------------------------------------
# helpers: metrics
# ----------------------------------------------------------------------
 
def _corr(a, b):
    """Pearson correlation with NaN handling. (ACC: constant climatological
    offset does not change correlation, so anomalies are implicit.)"""
    m = np.isfinite(a) & np.isfinite(b)
    if m.sum() < 3:
        return np.nan
    a = a[m] - a[m].mean()
    b = b[m] - b[m].mean()
    d = np.sqrt((a * a).sum() * (b * b).sum())
    return float((a * b).sum() / d) if d > 0 else np.nan
 
 
def crps_gaussian(obs, mu, sigma):
    z = (obs - mu) / sigma
    return sigma * (z * (2.0 * norm.cdf(z) - 1.0)
                    + 2.0 * norm.pdf(z) - 1.0 / np.sqrt(np.pi))
 
 
def crps_ensemble(obs, ens):
    """obs (...,), ens (..., M). Fair-ish PWM form: E|X-y| - 0.5 E|X-X'|."""
    import warnings
    with warnings.catch_warnings():
        warnings.simplefilter('ignore', category=RuntimeWarning)
        t1 = np.nanmean(np.abs(ens - obs[..., None]), axis=-1)
        t2 = np.nanmean(np.abs(ens[..., :, None] - ens[..., None, :]),
                        axis=(-2, -1))
    return t1 - 0.5 * t2
 
 
def crps_clim_reference(obs, clim_members):
    """CRPS of the fixed climatological ensemble (training years as members)."""
    c = np.asarray(clim_members, dtype=float)
    c = c[np.isfinite(c)]
    t2 = np.mean(np.abs(c[:, None] - c[None, :]))
    # c is NaN-free, so any NaN comes from obs; plain mean propagates it
    # silently instead of nanmean's "Mean of empty slice" RuntimeWarning
    t1 = np.mean(np.abs(c[None, None, :] - obs[..., None]), axis=-1)
    return t1 - 0.5 * t2
 
 
def pit_values(obs, src):
    """PIT: Gaussian CDF for parametric, randomized rank for ensemble."""
    if src['kind'] == 'gaussian':
        pit = norm.cdf(obs, src['mu'], src['sigma'])
    else:
        ens, M = src['ens'], src['ens'].shape[-1]
        less = np.sum(ens < obs[..., None], axis=-1)
        equal = np.sum(ens == obs[..., None], axis=-1)
        u = RNG.uniform(size=less.shape)
        pit = (less + u * (equal + 1)) / (M + 1)
    return pit[np.isfinite(obs) & np.isfinite(src['mu'])]
 
 
def exceed_prob(src, thresh, tail):
    """P(x > q90) for tail='upper', P(x < q10) for tail='lower'."""
    if src['kind'] == 'gaussian':
        cdf = norm.cdf(thresh, src['mu'], src['sigma'])
        return (1.0 - cdf) if tail == 'upper' else cdf
    ens = src['ens']
    if tail == 'upper':
        return np.nanmean(ens > thresh, axis=-1)
    return np.nanmean(ens < thresh, axis=-1)
 
 
def by_lead(arr2d, func=np.nanmean):
    """Reduce (init, lead) -> (lead,) with NaN-aware func."""
    import warnings
    with warnings.catch_warnings():
        warnings.simplefilter('ignore', category=RuntimeWarning)
        return np.array([func(arr2d[:, l]) for l in range(arr2d.shape[1])])
 
 
# ----------------------------------------------------------------------
# per-source metric computation
# ----------------------------------------------------------------------
 
def compute_metrics(var_, kind, ds, ds_target, ds_target_hist,
                    years_verif, common_inits=None):
    src = extract_source(var_, kind, ds, common_inits)
    obs, leads = align_obs(src['ds'], ds_target, var_, years_verif)
 
    hist = np.asarray(ds_target_hist[var_].values, dtype=float)
    hist = hist[np.isfinite(hist)]
    q10, q90 = np.nanquantile(hist, [0.10, 0.90])
 
    out = dict(leads=leads)
 
    # --- deterministic ---
    err = src['mu'] - obs
    out['acc'] = np.array([_corr(src['mu'][:, l], obs[:, l])
                           for l in range(len(leads))])
    out['rmse'] = np.sqrt(by_lead(err ** 2))
    out['spread'] = np.sqrt(by_lead(src['sigma'] ** 2))
    out['std_fcst'] = np.array([np.nanstd(np.where(np.isfinite(obs[:, l]),
                                                   src['mu'][:, l], np.nan), ddof=1)
                                for l in range(len(leads))])
    out['std_obs'] = np.array([np.nanstd(obs[:, l], ddof=1)
                               for l in range(len(leads))])
 
    # --- probabilistic ---
    if kind == 'gaussian':
        crps = crps_gaussian(obs, src['mu'], src['sigma'])
    else:
        crps = crps_ensemble(obs, src['ens'])
    crps = np.where(np.isfinite(obs), crps, np.nan)
    out['crps'] = by_lead(crps)
 
    crps_c = crps_clim_reference(obs, hist)
    crps_c = np.where(np.isfinite(obs), crps_c, np.nan)
    out['crpss'] = 1.0 - out['crps'] / by_lead(crps_c)
 
    out['pit'] = pit_values(obs, src)
 
    # --- Brier skill, q90 / q10 (climatological reference p = 0.1) ---
    for tag, thr, tail in (('q90', q90, 'upper'), ('q10', q10, 'lower')):
        p = exceed_prob(src, thr, tail)
        o = (obs > thr).astype(float) if tail == 'upper' else (obs < thr).astype(float)
        o = np.where(np.isfinite(obs), o, np.nan)
        bs = by_lead((p - o) ** 2)
        bs_clim = by_lead((0.10 - o) ** 2)
        out[f'bss_{tag}'] = 1.0 - bs / bs_clim
 
    return out
 
 
# ----------------------------------------------------------------------
# Taylor diagram axes
# ----------------------------------------------------------------------
 
def _setup_taylor(ax, rmax=1.8):
    corr_ticks = np.array([0.0, 0.2, 0.4, 0.6, 0.8, 0.9, 0.95, 0.99, 1.0])
    ax.set_thetalim(0, np.pi / 2)
    ax.set_theta_zero_location('E')
    ax.set_theta_direction(1)
    ax.set_thetagrids(np.degrees(np.arccos(corr_ticks)),
                      labels=[f'{c:g}' for c in corr_ticks], fontsize=8)
    ax.set_rmax(rmax)
    ax.set_rlabel_position(135)
    ax.tick_params(labelsize=8)
    th = np.linspace(0, np.pi / 2, 100)
    ax.plot(th, np.ones_like(th), 'k--', lw=0.8)            # std ratio = 1
    ax.plot(0, 1.0, 'k*', ms=12, zorder=5)                   # REF (obs)
    ax.text(np.pi / 4 * 1.6, rmax * 1.12, 'Correlation', fontsize=9,
            ha='center', rotation=0)
    ax.set_xlabel('Normalized std. dev.', fontsize=9)
 
 
def _taylor_point(corr, std_ratio):
    c = np.clip(corr, 0.0, 1.0)        # negative-corr points pinned at theta=90
    return np.arccos(c), std_ratio
 
 
# ----------------------------------------------------------------------
# main figure routine
# ----------------------------------------------------------------------
 
def verify_variable(var_, sources, ds_target, ds_target_hist, years_verif,
                    common_inits=None, figsize=(13, 14), title='TEST'):
    """
    sources : dict {label: (kind, dataset)} with kind in {'gaussian','ensemble'}.
              Put 'Raw' last so it is drawn in gray.
    """
    metrics = {}
    for label, (kind, ds) in sources.items():
        metrics[label] = compute_metrics(var_, kind, ds, ds_target,
                                         ds_target_hist, years_verif,
                                         common_inits=common_inits)
 
    labels = list(metrics.keys())
    cmap = plt.get_cmap('tab10')
    colors = {lab: ('0.35' if lab.lower().startswith('raw') else cmap(i))
              for i, lab in enumerate(labels)}
    leads = metrics[labels[0]]['leads']
    x = np.arange(len(leads))
 
    fig = plt.figure(figsize=figsize)
    gs = fig.add_gridspec(3, 2, hspace=0.32, wspace=0.25)
    ax1 = fig.add_subplot(gs[0, 0])
    axT = fig.add_subplot(gs[0, 1], projection='polar')
    ax3 = fig.add_subplot(gs[1, 0])
    ax4 = fig.add_subplot(gs[1, 1])
    ax5 = fig.add_subplot(gs[2, 0])
    ax6 = fig.add_subplot(gs[2, 1])
 
    # (1) ACC bars by lead
    w = 0.8 / len(labels)
    for k, lab in enumerate(labels):
        ax1.bar(x + (k - (len(labels) - 1) / 2) * w, metrics[lab]['acc'],
                width=w, color=colors[lab], label=lab)
    ax1.axhline(0, color='k', lw=0.8)
    ax1.set_xticks(x); ax1.set_xticklabels(leads)
    ax1.set_xlabel('lead_year'); ax1.set_ylabel('ACC (ensemble mean)')
    ax1.set_title('(a) ACC by lead year')
    ax1.legend(fontsize=8, ncol=2, frameon=False)
 
    # (2) Taylor diagram, one point per lead per source
    rmax = 1.05 * max(1.0, np.nanmax([np.nanmax(metrics[l]['std_fcst'] /
                                                metrics[l]['std_obs'])
                                      for l in labels]))
    _setup_taylor(axT, rmax=min(rmax, 2.5))
    for lab in labels:
        m = metrics[lab]
        for j, ld in enumerate(m['leads']):
            th, r = _taylor_point(m['acc'][j], m['std_fcst'][j] / m['std_obs'][j])
            if np.isfinite(th) and np.isfinite(r):
                axT.plot(th, min(r, axT.get_rmax() * 0.98), 'o',
                         color=colors[lab], ms=5, mec='none', alpha=0.85)
                axT.annotate(str(ld), (th, min(r, axT.get_rmax() * 0.98)),
                             fontsize=6, xytext=(2, 2),
                             textcoords='offset points')
    axT.set_title('(b) Taylor diagram (labels = lead_year)', pad=18)
 
    # (3) PIT histogram
    bins = np.linspace(0, 1, 11)
    for lab in labels:
        ax3.hist(metrics[lab]['pit'], bins=bins, density=True,
                 histtype='step', lw=1.8, color=colors[lab], label=lab)
    ax3.axhline(1.0, color='k', ls='--', lw=0.8)
    ax3.set_xlabel('PIT'); ax3.set_ylabel('Density')
    ax3.set_title('(c) PIT histogram (pooled)')
    ax3.legend(fontsize=8, frameon=False)
 
    # (4) RMSE (solid) and spread (dashed)
    for lab in labels:
        m = metrics[lab]
        ax4.plot(leads, m['rmse'], '-o', ms=4, color=colors[lab], label=lab)
        ax4.plot(leads, m['spread'], '--s', ms=3, color=colors[lab], alpha=0.7)
    ax4.set_xlabel('lead_year'); ax4.set_ylabel('RMSE (solid) / Spread (dashed)')
    ax4.set_title('(d) Ensemble-mean RMSE & spread')
    ax4.legend(fontsize=8, frameon=False)
 
    # (5) CRPS (left) and CRPSS (right, dashed)
    ax5b = ax5.twinx()
    for lab in labels:
        m = metrics[lab]
        ax5.plot(leads, m['crps'], '-o', ms=4, color=colors[lab], label=lab)
        ax5b.plot(leads, m['crpss'], '--s', ms=3, color=colors[lab], alpha=0.7)
    ax5b.axhline(0, color='k', lw=0.8, ls=':')
    ax5.set_xlabel('lead_year')
    ax5.set_ylabel('CRPS (solid)'); ax5b.set_ylabel('CRPSS vs clim. (dashed)')
    ax5.set_title('(e) CRPS & CRPSS by lead year')
    ax5.legend(fontsize=8, frameon=False)
 
    # (6) BSS q90 (solid) and q10 (dashed)
    for lab in labels:
        m = metrics[lab]
        ax6.plot(leads, m['bss_q90'], '-o', ms=4, color=colors[lab], label=lab)
        ax6.plot(leads, m['bss_q10'], '--s', ms=3, color=colors[lab], alpha=0.7)
    ax6.axhline(0, color='k', lw=0.8)
    ax6.set_xlabel('lead_year')
    ax6.set_ylabel('BSS  (solid: >q90, dashed: <q10)')
    ax6.set_title('(f) BSS for tail events (thresholds from 1968-1999)')
    ax6.legend(fontsize=8, frameon=False)
 
    fig.suptitle(title, fontsize=14, y=0.99)
    return fig, metrics
 
 
# ----------------------------------------------------------------------
# convenience: choose Gaussian (calib) vs ensemble (dress) per variable
# ----------------------------------------------------------------------
 
def pick_source(var_, ds_calib_x, ds_dress_x):
    """Gaussian closed forms for non-precip (var present in calib ds),
    member-based otherwise."""
    if f'{var_}_mu' in ds_calib_x.data_vars:
        return ('gaussian', ds_calib_x)
    return ('ensemble', ds_dress_x)
```
